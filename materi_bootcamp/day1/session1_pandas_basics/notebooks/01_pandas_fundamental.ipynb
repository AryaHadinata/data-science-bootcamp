## 1️⃣ Setup & Load Data

```python
import pandas as pd
import numpy as np
from pathlib import Path

# Konfigurasi display
pd.set_option('display.max_columns', 20)
pd.set_option('display.float_format', '{:.2f}'.format)

print(f"Pandas version: {pd.__version__}")
```

```python
# Load dataset
data_path = Path('../../../../datasets/StudentPerformanceFactors.csv')

df = pd.read_csv(data_path)

print(f"✅ Data loaded: {len(df):,} rows, {len(df.columns)} columns")
```

## 2️⃣ Eksplorasi Awal

```python
# Lihat 5 baris pertama
df.head()
```

```python
# Info struktur data
df.info()
```

```python
# Summary statistics untuk kolom numerik
df[['Hours_Studied']].describe()
```

## 3️⃣ Seleksi Data

```python
# Pilih kolom penting
key_cols = ['Hours_Studied', 'Sleep_Hours', 'Parental_Education_Level', 'Gender', 'Exam_Score']
df_simple = df[key_cols].copy()
df_simple.head()
```

```python
# Filter:exam score > 50
high_value = df_simple[df_simple['Exam_Score'] > 50]
print(f"Exam Score > 50: {len(high_value):,}")
high_value.head()
```

```python

tender_high = df_simple[
    (df_simple['Exam_Score'] > 50) & 
    (df_simple['Parental_Education_Level'] == 'High School')
]
print(f"Exam & Parental> 50: {len(tender_high):,}")
```

## 4️⃣ Agregasi & GroupBy

```python
# Hitung statistik ujian berdasarkan tingkat motivasi
performa_per_motivasi = df.groupby('Motivation_Level')['Exam_Score'].agg([
    ('jumlah_siswa', 'count'),
    ('total_skor', 'sum'),
    ('rata_rata_skor', 'mean')
]).round(2)

performa_per_motivasi.sort_values('rata_rata_skor', ascending=False)
```

```python
top_school_teacher = df.groupby(['School_Type', 'Teacher_Quality'])['Exam_Score'].agg([
    ('jumlah_siswa', 'count'),
    ('rata_rata_skor', 'mean')
]).sort_values('rata_rata_skor', ascending=False).head(10).round(2)

top_school_teacher
```

## 5️⃣ Missing Values

```python
# Cek missing values
missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(2)

missing_summary = pd.DataFrame({
    'missing_count': missing,
    'missing_pct': missing_pct
}).query('missing_count > 0').sort_values('missing_count', ascending=False)

missing_summary.head(10)
```

## 6️⃣ Sorting & Ranking

```python
top_exam = df.nlargest(10, 'Exam_Score')[
    ['Gender', 'School_Type', 'Hours_Studied', 'Previous_Scores', 'Exam_Score']
]
top_exam
```
